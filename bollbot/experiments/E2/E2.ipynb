```python
import sys
from pathlib import Path

here = Path.cwd()
root = next(p for p in [here, *here.parents] if (p / 'bollbot' / 'analysis').is_dir())
sys.path.insert(0, str(root))
from bollbot.analysis.report import ANALYSES

res = ANALYSES['E2'](root / 'bollbot' / 'experiments' / 'E2', root / 'bollbot' / 'experiments' / 'figures')
print(res.title, '->', res.status, '' if res.n is None else f'(n = {res.n})')
for k in res.kpis:
    print(f'  {k.label}: {k.value} (target {k.target_text}) -> {k.status} {k.note}')
for note in res.notes:
    print('  -', note)
```

```python
import pandas as pd  # tables (pip install pandas)
from IPython.display import Image, display

for name, rows in res.tables.items():
    print(name)
    display(pd.DataFrame(rows))
for fig in res.figures:
    display(Image(filename=str(fig)))
```
